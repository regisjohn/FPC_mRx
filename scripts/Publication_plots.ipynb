```python
import numpy as np
import matplotlib.pyplot as plt
import os
import h5py
import fipcore as fpc
import pyspedas
from pyspedas.projects import mms
from pyspedas import get_data, tplot_rename, tplot_restore
```

```python
probe = '1'
trange = ['2015-12-09/05:03:55', '2015-12-09/05:03:59']
data_rate = 'brst'
level = 'l2'
species = 'e'
subtract_f0 = True
type_tag = 'df' if subtract_f0 else 'f'
bin_width_frac = 0.25
project_root = os.path.dirname(os.path.dirname(os.path.dirname(__file__))) if '__file__' in locals() else ".."
data_dir = os.path.join(project_root, "data")
plot_dir = os.path.join(project_root, "plots")
get_support_data = True
no_update = True
hfile_pref = f'mms{probe}_{data_rate}_{type_tag}_{species}_{bin_width_frac:.2f}'
hfile = os.path.join(data_dir, fpc.mms_name_make(hfile_pref, trange[0], trange[1]))
```

```python
hfile
```

```python
fpi_vars = mms.fpi(trange= trange, probe=probe, data_rate=data_rate, level=level,
    datatype=['des-dist', 'dis-dist', 'des-moms', 'dis-moms'], 
    time_clip=True, varnames=[f'mms{probe}_des_energy_brst', f'mms{probe}_des_phi_brst', 
    f'mms{probe}_des_bulkv_gse_brst', f'mms{probe}_dis_bulkv_gse_brst', 
    f'mms{probe}_des_temppara_brst', f'mms{probe}_des_tempperp_brst', 
    f'mms{probe}_des_numberdensity_brst', f'mms{probe}_des_dist_brst', 
    f'mms{probe}_des_disterr_brst', f'mms{probe}_des_errorflags_brst_dist'], 
    get_support_data=get_support_data, no_update=no_update)
```

```python
# Read in EDP data
edp_vars = mms.edp(trange=trange, probe=probe, data_rate=data_rate, level=level,
    datatype=['dce', 'scpot'],
    time_clip=True, varnames=[f'mms{probe}_edp_scpot_brst_l2', 
    f'mms{probe}_edp_dce_gse_brst_l2', f'mms{probe}_edp_dce_par_epar_brst_l2'], 
    get_support_data=get_support_data, no_update=no_update)
```

```python
# Read in FGM data
fgm_vars = mms.fgm(trange=trange, probe=probe, data_rate=data_rate, level=level,
    varnames=f'mms{probe}_fgm_b_gse_brst_l2', time_clip=True, 
    get_support_data=get_support_data, no_update=no_update)
```

```python
with h5py.File(hfile, "r") as f:
    ntime_e = f["meta"]["time"][...]
    bvec_lmn = f["lmn"]["bvec"][...]
    evec_lmn = f["lmn"]["evec_lor"][...]
    evec_fac = f["fac"]["evec_lor"][...]
    jvec_lmn = f["lmn"]["jvec"][...]
    jvec_fac = f["fac"]["jvec"][...]
    jdotE_fac = f["fac"]["jdotE"][...]
    jdotE_lmn = f["lmn"]["jdotE"][...]
    fac_mat = f["fac"]['rot_mat'][...]
    lmn_mat = f["lmn"]['rot_mat'][...]
    den_e = f['meta']['den_e'][...]
    den_i = f['meta']['den_i'][...]
    te_para = f['meta']['te_para'][...]
    te_perp = f['meta']['te_perp'][...]
    ti_para = f['meta']['ti_para'][...]
    ti_perp = f['meta']['ti_perp'][...]
    bulk_ve_lmn = f["lmn"]['bulk_ve'][...]
    bulk_ve_fac = f['fac']['bulk_ve'][...]
    bulk_vi_lmn = f["lmn"]['bulk_vi'][...]
    bulk_vi_fac = f["fac"]['bulk_vi'][...]
    bvec_lmn = f['lmn']['bvec'][...]
    evec_lmn = f['lmn']['evec_lor'][...]
    evec_fac = f['fac']['evec_lor'][...]
    jvec_lmn = f['lmn']['jvec'][...]
    jvec_fac = f['fac']['jvec'][...]
    jdotE_fac = f['fac']['jdotE'][...]
    jdotE_lmn = f['lmn']['jdotE'][...]
    bmag = f['meta']['bmag'][...]
```

```python
from pyspedas import time_datetime

t_ax = time_datetime(ntime_e)
axs_label_fntsz = 16   # adjust as needed
legend_fntsz = 14
legendFrame = True

n_subplots = 16

fig, ax = plt.subplots(n_subplots, 1, figsize=(12, 28),  sharex=True, constrained_layout=True)

# Figure title:
fig.suptitle(f"MMS {probe}: {trange[0]} – {trange[1]}", fontsize=18, fontweight='bold')

# First two subplots: energy spectra of electrons and ions
fig, ax0 = pyspedas.tplot('energy_e', fig = fig, axis = ax[0], display=False, return_plot_objects=True)
fig, ax1 = pyspedas.tplot('energy_i', fig = fig, axis = ax[1], display=False, return_plot_objects=True)
ax0.set_ylabel(r"$Electrons$" "\n" r"$(\mathrm{eV})$", fontsize=axs_label_fntsz)
ax1.set_ylabel(r"$Ions$" "\n" r"$(\mathrm{eV})$", fontsize=axs_label_fntsz)

# Third subplot: density of electrons and ions
ax[2].set_prop_cycle(color=['blue', 'red']) # custom color cycle
ax[2].plot(t_ax, den_e, linewidth=1.5, label=r'$n_e$')
ax[2].plot(t_ax, den_i, linewidth=1.5, label=r'$n_i$')
ax[2].axhline(0, color='k', linestyle='dotted', linewidth=1.0)
ax[2].set_ylabel(r"$n$" "\n" r"$(\mathrm{cm^{-3}})$", fontsize=axs_label_fntsz)
ax[2].legend(fontsize=legend_fntsz, frameon=legendFrame)

# Fourth subplot: electron temp profiles
ax[3].set_prop_cycle(color=['red', 'blue']) # custom color cycle
ax[3].plot(t_ax, te_para, linewidth=1.5, label=r'$T_{\parallel e}$')
ax[3].plot(t_ax, te_perp, linewidth=1.5, label=r'$T_{\perp e}$')
ax[3].axhline(0, color='k', linestyle='dotted', linewidth=1.0)
ax[3].set_ylabel(r"$T_e$" "\n" r"$(\mathrm{eV})$", fontsize=axs_label_fntsz)
ax[3].legend(fontsize=legend_fntsz, frameon=legendFrame)

# Fifth subplot: ion temp profiles
ax[4].set_prop_cycle(color=['red', 'blue']) # custom color cycle
ax[4].plot(t_ax, ti_para, linewidth=1.5, label=r'$T_{\parallel i}$')
ax[4].plot(t_ax, ti_perp, linewidth=1.5, label=r'$T_{\perp i}$')
ax[4].axhline(0, color='k', linestyle='dotted', linewidth=1.0)
ax[4].set_ylabel(r"$T_i$" "\n" r"$(\mathrm{eV})$", fontsize=axs_label_fntsz)
ax[4].legend(fontsize=legend_fntsz, frameon=legendFrame)

# Sixth subplot: electron bulk flow in lmn
ax[5].set_prop_cycle(color=['blue', 'green', 'red']) # custom color cycle
ax[5].plot(t_ax, bulk_ve_lmn, linewidth=1.5, label=["$U_{eL}$", "$U_{eM}$", "$U_{eN}$"])
ax[5].axhline(0, color='k', linestyle='dotted', linewidth=1.0)
ax[5].set_ylabel("$U_{eMVA}$" "\n" r"$(\mathrm{km/s})$", fontsize=axs_label_fntsz)
ax[5].legend(fontsize=legend_fntsz, frameon=legendFrame, loc='upper right')

# Seventh subplot: electron bulk flow in fac
ax[6].set_prop_cycle(color=['blue', 'green', 'red']) # custom color cycle
ax[6].plot(t_ax, bulk_ve_fac, linewidth=1.5, label=[r"$U_{e\perp_1}$", r"$U_{e\perp_2}$", 
                                                    r"$U_{e\parallel}$"])
ax[6].axhline(0, color='k', linestyle='dotted', linewidth=1.0)
ax[6].set_ylabel("$U_{eFAC}$" "\n" r"$(\mathrm{km/s})$", fontsize=axs_label_fntsz)
ax[6].legend(fontsize=legend_fntsz, frameon=legendFrame, loc='upper right')

# Eighth subplot: ion bulk flow in lmn
ax[7].set_prop_cycle(color=['blue', 'green', 'red']) # custom color cycle
ax[7].plot(t_ax, bulk_vi_lmn, linewidth=1.5, label=["$U_{iL}$", "$U_{iM}$", "$U_{iN}$"])
ax[7].axhline(0, color='k', linestyle='dotted', linewidth=1.0)
ax[7].set_ylabel("$U_{iMVA}$" "\n" r"$(\mathrm{km/s})$", fontsize=axs_label_fntsz)
ax[7].legend(fontsize=legend_fntsz, frameon=legendFrame, loc='upper right')

# Ninth subplot: ion bulk flow in fac
ax[8].set_prop_cycle(color=['blue', 'green', 'red']) # custom color cycle
ax[8].plot(t_ax, bulk_vi_fac, linewidth=1.5, label=[r"$U_{i\perp_1}$", r"$U_{i\perp_2}$", 
                                                    r"$U_{i\parallel}$"])
ax[8].axhline(0, color='k', linestyle='dotted', linewidth=1.0)
ax[8].set_ylabel(r"$U_{iFAC}$" "\n" r"$(\mathrm{km/s})$", fontsize=axs_label_fntsz)
ax[8].legend(fontsize=legend_fntsz, frameon=legendFrame, loc='upper right')

# Tenth subplot: B field in lmn
ax[9].set_prop_cycle(color=['blue', 'green', 'red', 'black']) # custom color cycle
ax[9].plot(t_ax, bvec_lmn, linewidth=1.5, label=['$B_L$', '$B_M$', '$B_N$'])
ax[9].plot(t_ax, bmag, linewidth=1.5, label='$|B|$')
ax[9].axhline(0, color='k', linestyle='dotted', linewidth=1.0)
ax[9].set_ylabel(r"$B$" "\n" r"$(\mathrm{nT})$", fontsize=axs_label_fntsz)
ax[9].legend(fontsize=legend_fntsz, frameon=legendFrame)

# Eleventh subplot: E field in LMN coordinates
ax[10].set_prop_cycle(color=['blue', 'green', 'red']) # custom color cycle
ax[10].plot(t_ax, evec_lmn, linewidth=1.5, label=["$E_L$", "$E_M$", "$E_N$"])
ax[10].axhline(0, color='k', linestyle='dotted', linewidth=1.0)
ax[10].set_ylabel(r"$E_{MVA}$" "\n" r"$(\mathrm{mV/m})$", fontsize=axs_label_fntsz)
ax[10].legend(fontsize=legend_fntsz, frameon=legendFrame, loc='upper right')

# Twelfth subplot: E field in FAC coordinates
ax[11].set_prop_cycle(color=['blue', 'green', 'red']) # custom color cycle
ax[11].plot(t_ax, evec_fac, linewidth=1.5, label=[r"$E_{\perp_1}$", r"$E_{\perp_2}$", 
                                                  r"$E_{\parallel}$"])
ax[11].axhline(0, color='k', linestyle='dotted', linewidth=1.0)
ax[11].set_ylabel("$E_{FAC}$" "\n" r"$(\mathrm{mV/m})$", fontsize=axs_label_fntsz)
ax[11].legend(fontsize=legend_fntsz, frameon=legendFrame, loc='upper right')

# Thirteenth subplot: J vector in lmn coordinates
ax[12].set_prop_cycle(color=['blue', 'green', 'red']) # custom color cycle
ax[12].plot(t_ax, jvec_lmn*1e-3, linewidth=1.5, label=["$J_L$", "$J_M$", "$J_N$"])
ax[12].axhline(0, color='k', linestyle='dotted', linewidth=1.0)
ax[12].set_ylabel(r"$J_{MVA}$" "\n" r"$(\mu\mathrm{A}/\mathrm{m}^2)$", fontsize=axs_label_fntsz)
ax[12].legend(fontsize=legend_fntsz, frameon=legendFrame, loc='upper right')


# Fourteenth subplot: J vector in fac coordinates
ax[13].set_prop_cycle(color=['blue', 'green', 'red']) # custom color cycle
ax[13].plot(t_ax, jvec_fac*1e-3, linewidth=1.5, label=[r"$J_{\perp_1}$", r"$J_{\perp_2}$", 
                                                       r"$J_{\parallel}$"])
ax[13].axhline(0, color='k', linestyle='dotted', linewidth=1.0)
ax[13].set_ylabel(r"$J_{FAC}$" "\n" r"$(\mu\mathrm{A}/\mathrm{m}^2)$", fontsize=axs_label_fntsz)
ax[13].legend(fontsize=legend_fntsz, frameon=legendFrame, loc='upper right')

# Fifteenth subplot: J.E in lmn coordinates
ax[14].set_prop_cycle(color=['black', 'blue', 'green', 'red']) # custom color cycle
ax[14].plot(t_ax, jdotE_lmn, linewidth=1.5, 
    label=[r"$J \cdot E_{tot}$", r"$J \cdot E_{L}$", r"$J \cdot E_{M}$", r"$J \cdot E_{N}$"])
ax[14].axhline(0, color='k', linestyle='dotted', linewidth=1.0)
ax[14].set_ylabel(r"$J \cdot E_{MVA}$" "\n" r"$(\mathrm{nW/m^3})$", fontsize=axs_label_fntsz)
ax[14].legend(fontsize=legend_fntsz, loc='upper left')

# Sixteenth subplot: J.E in fac coordinates
ax[15].set_prop_cycle(color=['black', 'blue', 'green', 'red']) # custom color cycle
ax[15].plot(t_ax, jdotE_fac, linewidth=1.5, 
    label=[r"$J·E_{tot}$", r"$J·E_{\perp_1}$", r"$J·E_{\perp_2}$", r"$J·E_{\parallel}$"])
ax[15].axhline(0, color='k', linestyle='dotted', linewidth=1.0)
ax[15].set_ylabel(r"$J \cdot E_{FAC}$" "\n" r"$(\mathrm{nW/m^3})$", fontsize=axs_label_fntsz)
ax[15].legend(fontsize=legend_fntsz, loc='upper left')


# Vertical reference lines
y_up = n_subplots + 1.36
ax[-1].axvline(x=dt_obj, ymin=0, ymax=y_up, color='k', linestyle='dashed', 
    linewidth=1, clip_on=False)
ax[-1].axvline(x=dt_obj0, ymin=0, ymax=y_up, color='k', linestyle='dotted', 
    linewidth=1, clip_on=False)
ax[-1].axvline(x=dt_obj1, ymin=0, ymax=y_up, color='k', linestyle='dotted', 
    linewidth=1, clip_on=False)

# Save routine
plt_pref = f'mms{probe}_{data_rate}_{species}_{bin_width_frac:.2f}'
pltfile = os.path.join(plot_dir, fpc.mms_name_make(plt_pref, 
                                                   trange[0], trange[1], ext='pdf'))
fig.savefig(pltfile, bbox_inches='tight', dpi=300)
print(f'File saved to {pltfile}!')

plt.show()
```

```python
import numpy as np
import h5py

def load_plt_mms_data(hfiles, time_indices):
    """
    Load FPC data for all probes and return integrated perp1, perp2, and parallel
    components for both C_E and dvdf, at a given time_index.
    """
    # If time_indices is just one int, convert to a list
    if isinstance(time_indices, int):
        time_indices = [time_indices] * len(hfiles)

    probes = []

    for hfile, time_index in zip(hfiles, time_indices):
        with h5py.File(hfile, "r") as f:
            c_vol = f["fac"]["c_vol"][...]          # (3,28,28,28,133)
            vpar   = f["fac"]["binc_n"][...]        # (28,)
            dvdf   = f["fac"]["bvdf_vol"][...]      # (28,28,28,133)
            tarr   = f["meta"]["time"][...]         # (133,)

        # Slice the 3 components of fpc and delta f at time_index
        c_perp1 = c_vol[0, :, :, :, time_index]
        c_perp2 = c_vol[1, :, :, :, time_index]
        c_par   = c_vol[2, :, :, :, time_index]

        dvdf_t = dvdf[:, :, :, time_index]

        # Integrate over two directions
        # C_perp1
        c_perp1_vperp1 = np.nansum(c_perp1, axis=(1, 2))
        c_perp1_vperp2 = np.nansum(c_perp1, axis=(0, 2))
        c_perp1_vpar   = np.nansum(c_perp1, axis=(0, 1))

        # C_perp2
        c_perp2_vperp1 = np.nansum(c_perp2, axis=(1, 2))
        c_perp2_vperp2 = np.nansum(c_perp2, axis=(0, 2))
        c_perp2_vpar   = np.nansum(c_perp2, axis=(0, 1))

        # C_parallel
        c_par_vperp1 = np.nansum(c_par, axis=(1, 2))
        c_par_vperp2 = np.nansum(c_par, axis=(0, 2))
        c_par_vpar   = np.nansum(c_par, axis=(0, 1))

        # delta vdf integration
        dvdf_vperp1 = np.nansum(dvdf_t, axis=(1, 2))
        dvdf_vperp2 = np.nansum(dvdf_t, axis=(0, 2))
        dvdf_vpar   = np.nansum(dvdf_t, axis=(0, 1))

        probes.append({

            "vpar": vpar,

            "c_e": {
                "c_perp1": {
                    "vperp1": c_perp1_vperp1,
                    "vperp2": c_perp1_vperp2,
                    "vpar":   c_perp1_vpar,
                },

                "c_perp2": {
                    "vperp1": c_perp2_vperp1,
                    "vperp2": c_perp2_vperp2,
                    "vpar":   c_perp2_vpar,
                },

                "c_par": {
                    "vperp1": c_par_vperp1,
                    "vperp2": c_par_vperp2,
                    "vpar":   c_par_vpar,
                },
            },
             "df": {
                "vperp1": dvdf_vperp1,
                "vperp2": dvdf_vperp2,
                "vpar":   dvdf_vpar,
            },

            "time": tarr[time_index]
        })

    return probes
```

```python
import matplotlib.pyplot as plt
from pyspedas import time_string


def plt_vdf_fpc(trange, probes_data, time_index, prb=0, figsize=(18, 14), 
                save_fig=False, sav_ext='png'):
    """
    4×3 subplot layout:

        Row 0: delta f
        Row 1: C_E perpendicular 1
        Row 2: C_E perpendicular 2
        Row 3: C_E parallel
    """
    # Velocity coordinate
    vpar = probes_data[0]["vpar"]

    fig, axs = plt.subplots(4, 3, figsize=figsize, constrained_layout=True)

    # Style for MMS1–4
    colors = ['red', 'green', 'blue', 'black']
    styles = ['-', '--', ':', '-.']
    zorders = [5, 4, 3, 2]

    # Velocity keys for the three columns
    v_keys = ['vperp1', 'vperp2', 'vpar']

    # Plot each probe
    for i, pdata in enumerate(probes_data):

        lbl = f"MMS{i+1}"
        color = colors[i]
        ls = styles[i]
        zo = zorders[i]

        for col, v_key in enumerate(v_keys):
            # Row 0: delta f
            axs[0, col].plot(vpar, pdata["df"][v_key], color=color, 
                    linestyle=ls, linewidth=2, zorder=zo, label=lbl)

            # Row 1: C_E perpendicular 1
            axs[1, col].plot(vpar, pdata["c_e"]["c_perp1"][v_key], color=color, 
                    linestyle=ls, linewidth=2, zorder=zo, label=lbl)

            # Row 2: C_E perpendicular 2
            axs[2, col].plot(vpar, pdata["c_e"]["c_perp2"][v_key], color=color, 
                    linestyle=ls, linewidth=2, zorder=zo, label=lbl)

            # Row 3: C_E parallel
            axs[3, col].plot(vpar, pdata["c_e"]["c_par"][v_key], color=color, 
                    linestyle=ls, linewidth=2, zorder=zo, label=lbl)

    # Axis formatting
    for ax in axs.flat:
        ax.set_xlim([-3, 3])
        ax.tick_params(labelsize=16)
        ax.axhline(0, color='black', linewidth=1, linestyle='dashed')
        ax.axvline(0, color='black', linewidth=1, linestyle='dashed')
        ax.ticklabel_format(style='sci', axis='y', scilimits=(0, 0), useMathText=True)
        ax.yaxis.get_offset_text().set_fontsize(14)

    # X-axis labels
    x_labels = [r'$v_{\perp 1}/v_{te}$', r'$v_{\perp 2}/v_{te}$', 
                r'$v_{\parallel}/v_{te}$']

    for i in range(3):
        axs[3, i].set_xlabel(x_labels[i], fontsize=20)

    # Y-axis labels
    axs[0, 0].set_ylabel(r'$\delta f\ (\mathrm{m^{-3}})$', fontsize=20)

    axs[1, 0].set_ylabel(r'$C_{\mathrm{E}\perp 1}\ (\mathrm{W\cdot m^{-3}})$', 
        fontsize=20)

    axs[2, 0].set_ylabel(r'$C_{\mathrm{E}\perp 2}\ (\mathrm{W\cdot m^{-3}})$', 
        fontsize=20)

    axs[3, 0].set_ylabel(r'$C_{\mathrm{E}\parallel}\ (\mathrm{W\cdot m^{-3}})$', 
        fontsize=20)

    # Legend
    handles, labels = axs[0, 0].get_legend_handles_labels()

    fig.legend(handles, labels, loc='upper center', 
        bbox_to_anchor=(0.5, 0.01), ncol=4, fontsize=16, frameon=True)

    # Time annotation
    fig.text(0.98, 0.98, f"t: {time_index}", ha="right", va="top", fontsize=18)

    fig.suptitle(f"MMS: {time_string(probes_data[prb]['time'])[:-3]}", 
        fontsize=18, fontweight='bold')

    # Save / show
    if save_fig:
        plt_pref = (f'1D_{time_index}_mms_{data_rate}_{species}_{bin_width_frac:.2f}')
        pltfile = os.path.join(plot_dir, fpc.mms_name_make(plt_pref, trange[0], 
            trange[1], ext=sav_ext))
        fig.savefig(pltfile, bbox_inches='tight', dpi=300)
        print(f'File saved to {pltfile}!')
        plt.close(fig)
    else:
        plt.show()
```

## Testing the window size for averaging

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Set random seed for perfectly reproducible results
np.random.seed(42)

# Create a single long time series array (t) with 1000 indices
total_indices = 1000
baseline = 1.0
peak_center = 500       # Put the peak exactly in the middle of our time series
peak_duration = 7       # Peak is 7 indices wide
peak_multiplier = 15    # Peak value is ~75.0

# 1. Generate the single time series data (t)
t = baseline + np.random.normal(loc=0.0, scale=0.2, size=total_indices)
half_duration = peak_duration // 2
peak_start = peak_center - half_duration
peak_end = peak_start + peak_duration

# Inject the single peak into this specific time series
t[peak_start:peak_end] = (baseline * peak_multiplier) + np.random.normal(0, 1.0, peak_duration)

# 2. Set up the plotting layout
plt.figure(figsize=(14, 7))
indices = np.arange(total_indices)

# Plot the raw time series data
plt.plot(indices, t, color='#7f8c8d', alpha=0.6, label='Time Series Signal (t)', zorder=1)

# Define windows and colors for the visual bounds
window_widths = [100, 200, 500]
colors = ['#e74c3c', '#e67e22', '#2ecc71'] # Red, Orange, Green
alphas = [0.15, 0.10, 0.05]

# Plot each window boundary centered at the peak
for width, color, alpha in zip(window_widths, colors, alphas):
    half_width = width // 2
    w_start = peak_center - half_width
    w_end = peak_center + half_width
    
    # Calculate window data to show the mean inside the legend label
    window_mean = np.mean(t[w_start:w_end])
    
    # Add a shaded region representing the window span
    plt.axvspan(w_start, w_end, color=color, alpha=alpha, 
                label=f'{width}-Index Window (Mean: {window_mean:.2f})', zorder=0)
    
    # Add dashed vertical lines to mark boundaries clearly
    plt.axvline(w_start, color=color, linestyle='--', alpha=0.5, zorder=2)
    plt.axvline(w_end, color=color, linestyle='--', alpha=0.5, zorder=2)

# Label the chart elements
plt.title('Single Peak Time Series vs. Centered Averaging Windows', fontsize=14, fontweight='bold')
plt.xlabel('Index (t)', fontsize=12)
plt.ylabel('Signal Value', fontsize=12)
plt.grid(True, linestyle=':', alpha=0.6)

# Limit X-axis to focus closely around the windows of interest
plt.xlim(200, 800)
plt.ylim(0, 85)

# Place the legend
plt.legend(loc='upper right', frameon=True, facecolor='white', edgecolor='none', shadow=True)

# Add text callout for the peak maximum
peak_max_val = np.max(t[peak_start:peak_end])
plt.annotate(f'Peak Max (~{peak_max_val:.1f})', xy=(peak_center, peak_max_val), 
             xytext=(peak_center + 25, peak_max_val - 5),
             arrowprops=dict(facecolor='black', shrink=0.05, width=1, headwidth=6))

# Render graph
plt.tight_layout()
plt.show()
```

## Upstream Plasma Parameters

```python
import os
import h5py
import numpy as np
from pyspedas.projects import mms
from pyspedas import get_data, del_data
from scipy.constants import m_p, m_e, e, mu_0, epsilon_0, c

def compute_upstream_params(trange_upstream, probe='1', data_rate='brst', 
    level='l2', species='e', bin_width_frac=0.25, get_support_data=True, 
    no_update=True):

    # Clean workspace
    del_data('*')

    # 1. Load FPI and FGM data for the upstream window
    fpi_varlist = [
    f'mms{probe}_des_numberdensity_brst', f'mms{probe}_dis_numberdensity_brst', 
    f'mms{probe}_des_temppara_brst', f'mms{probe}_dis_temppara_brst', 
    f'mms{probe}_des_tempperp_brst', f'mms{probe}_dis_tempperp_brst', 
    f'mms{probe}_des_bulkv_gse_brst', f'mms{probe}_dis_bulkv_gse_brst'
    ]
    
    mms.fpi(trange=trange_upstream, probe=probe, data_rate=data_rate, level=level, 
        datatype=['des-moms', 'dis-moms'], time_clip=True, 
        varnames=fpi_varlist, get_support_data=get_support_data, no_update=no_update)

    mms.fgm(trange=trange_upstream, probe=probe, data_rate=data_rate, level=level,
        varnames=f'mms{probe}_fgm_b_gse_brst_l2', time_clip=True, 
        get_support_data=get_support_data, no_update=no_update)

    # Extract the numpy arrays
    _, b_vec = get_data(f'mms{probe}_fgm_b_gse_brst_l2_bvec')
    _, bmag = get_data(f'mms{probe}_fgm_b_gse_brst_l2_btot')
    _, den_e = get_data(f'mms{probe}_des_numberdensity_brst')
    _, den_i = get_data(f'mms{probe}_dis_numberdensity_brst')
    _, te_para = get_data(f'mms{probe}_des_temppara_brst')
    _, ti_para = get_data(f'mms{probe}_dis_temppara_brst')
    _, te_perp = get_data(f'mms{probe}_des_tempperp_brst')
    _, ti_perp = get_data(f'mms{probe}_dis_tempperp_brst')
    _, ue_gse = get_data(f'mms{probe}_des_bulkv_gse_brst')
    _, ui_gse = get_data(f'mms{probe}_dis_bulkv_gse_brst')

    # Guide Field Ratio
    bg_ratio = 0.5

    # Computing averages
    b_vec_mean = np.nanmean(b_vec, axis=0)   # shape: (3,) in nT
    bmag_mean = np.nanmean(bmag)             # nT

    den_e_mean = np.nanmean(den_e)           # cm^-3
    den_i_mean = np.nanmean(den_i)           # cm^-3

    te_para_mean = np.nanmean(te_para)       # eV
    te_perp_mean = np.nanmean(te_perp)       # eV
    te_mean = (te_para_mean + 2.0 * te_perp_mean) / 3.0

    ti_para_mean = np.nanmean(ti_para)       # eV
    ti_perp_mean = np.nanmean(ti_perp)       # eV
    ti_mean = (ti_para_mean + 2.0 * ti_perp_mean) / 3.0

    ue_gse_mean = np.nanmean(ue_gse, axis=0) # shape: (3,) in km/s
    ui_gse_mean = np.nanmean(ui_gse, axis=0) # shape: (3,) in km/s

    # Convert mean values to SI units
    B0_T = bmag_mean * 1e-9                  # T
    n_e_m3 = den_e_mean * 1e6                # m^-3
    n_i_m3 = den_i_mean * 1e6                # m^-3
    Te_J = te_mean * e                       # J
    Ti_J = ti_mean * e                       # J

    # 1. Temperature Ratio & Plasma Betas
    ti_te_ratio = ti_mean / te_mean
    beta_e = (2.0 * mu_0 * n_e_m3 * Te_J) / (B0_T**2)
    beta_i = (2.0 * mu_0 * n_i_m3 * Ti_J) / (B0_T**2)

    # 2. Speeds in km/s
    vA = (B0_T / np.sqrt(mu_0 * n_i_m3 * m_p)) / 1e3    # km/s
    vti = (np.sqrt(2.0 * Ti_J / m_p)) / 1e3             # km/s
    vte = (np.sqrt(2.0 * Te_J / m_e)) / 1e3             # km/s
    # Ion sound speed (using electron temperature and proton mass)
    c_s = (np.sqrt(Te_J / m_p)) / 1e3                   # km/s

    # 3. Frequencies in Hz
    omega_ci = (e * B0_T) / m_p
    omega_ce = (e * B0_T) / m_e
    omega_pi = np.sqrt((n_i_m3 * e**2) / (m_p * epsilon_0))
    omega_pe = np.sqrt((n_e_m3 * e**2) / (m_e * epsilon_0))

    f_ci = omega_ci / (2.0 * np.pi)          # Hz
    f_ce = omega_ce / (2.0 * np.pi)          # Hz
    f_pi = omega_pi / (2.0 * np.pi)          # Hz
    f_pe = omega_pe / (2.0 * np.pi)          # Hz

    # 4. Length Scales (km)
    rho_i = (vti * 1e3 / omega_ci) / 1e3     # km
    rho_e = (vte * 1e3 / omega_ce) / 1e3     # km
    d_i = (c / omega_pi) / 1e3               # km
    d_e = (c / omega_pe) / 1e3               # km
    # Ion sound gyroradius
    rho_s = (c_s * 1e3 / omega_ci) / 1e3     # km

    lambda_D = np.sqrt((epsilon_0 * Te_J) / (n_e_m3 * e**2)) # m

    # --- Print Results ---
    print("\n" + "="*70)
    print("UPSTREAM PLASMA PARAMETERS")
    print("="*70)

    print("\n--- Raw Plasma Parameters ---")
    print(f"Average B_vec (nT): {b_vec_mean}")
    print(f"Average |B| (nT): {bmag_mean:.3f}")
    print(f"Average Electron Density (cm^-3): {den_e_mean:.3f}")
    print(f"Average Ion Density (cm^-3): {den_i_mean:.3f}")
    print(f"Average Te (para, perp, scalar) (eV): {te_para_mean:.3f}, {te_perp_mean:.3f}, {te_mean:.3f}")
    print(f"Average Ti (para, perp, scalar) (eV): {ti_para_mean:.3f}, {ti_perp_mean:.3f}, {ti_mean:.3f}")
    
    print("\n--- Derived Plasma Parameters ---")
    print(f"Beta Electron (beta_e): {beta_e:.4f}")
    print(f"Beta Ion (beta_i): {beta_i:.4f}")
    print(f"Ti / Te Ratio: {ti_te_ratio:.3f}")
    print(f"b_g = {bg_ratio:.3f}")

    print("\n--- Characteristic Velocities ---")
    print(f"Average Ue_gse (km/s): {ue_gse_mean}")
    print(f"Average Ui_gse (km/s): {ui_gse_mean}")
    print(f"Alfvén Speed (vA): {vA:.2f} km/s")
    print(f"Ion Thermal Speed (vti): {vti:.2f} km/s")
    print(f"Electron Thermal Speed (vte): {vte:.2f} km/s")
    print(f"Ion Sound Speed (c_s): {c_s:.2f} km/s")

    print("\n--- Characteristic Lengths ---")
    print(f"Ion Larmor Radius (rho_i): {rho_i:.2f} km")
    print(f"Electron Larmor Radius (rho_e): {rho_e:.2f} km")
    print(f"Ion Skin Depth (d_i): {d_i:.2f} km")
    print(f"Electron Skin Depth (d_e): {d_e:.2f} km")
    print(f"Ion Sound Gyroradius (rho_s): {rho_s:.2f} km")
    print(f"Debye Length (lambda_D): {lambda_D:.2f} m")

    print("\n--- Characteristic Frequencies ---")
    print(f"Ion Cyclotron Frequency (f_ci): {f_ci:.4f} Hz")
    print(f"Electron Cyclotron Frequency (f_ce): {f_ce:.2f} Hz")
    print(f"Ion Plasma Frequency (f_pi): {f_pi:.2f} Hz")
    print(f"Electron Plasma Frequency (f_pe): {f_pe:.2f} Hz")

    print("="*70)
```

```python
time_index = 62
with h5py.File(hfile_df, 'r') as f:
    c_vol = f["fac"]["c_vol"][:, :, :, :, time_index]
    xc = f["fac"]["binc_n"][...]
    yc = f["fac"]["binc_n"][...]
    zc = f["fac"]["binc_n"][...]
    ntime = f["meta"]["time"][...]
```

```python
with h5py.File(hfile_f, 'r') as f:
    cf_vol = f["fac"]["c_vol"][:, :, :, :, time_index]
    xc = f["fac"]["binc_n"][...]
    yc = f["fac"]["binc_n"][...]
    zc = f["fac"]["binc_n"][...]
    cz_folds = f["fac"]["cz_folds"][:,:,:, time_index]
```

```python
cfz_xy, cfz_yz, cfz_xz = fpc.slice3d_to_2d(cf_vol[2]) # first row full f
cfz_2d_list = (cfz_xy, cfz_xz, cfz_yz.T) # The order is xy, xz and zy 

cfoldz_2d_list = (cfz_xy, cz_folds[0, :, :], cz_folds[1, :, :]) # second row fold f

cz_xy, cz_yz, cz_xz = fpc.slice3d_to_2d(c_vol[2]) # third row with delta f
cz_2d_list = (cz_xy, cz_xz, cz_yz.T) # The order is xy, xz and zy 
```

```python
ax_pairs = ((xc, yc), (xc, zc), (zc, yc))
ax_lbls = [r"$v_{\perp 1}/v_{te}$", r"$v_{\perp 2}/v_{te}$", 
                           r"$v_{\parallel}/v_{te}$"]
axlabel_pairs = ((ax_lbls[0], ax_lbls[1]), (ax_lbls[0], ax_lbls[2]), 
                 (ax_lbls[2], ax_lbls[1])) # xy, xz, zy
row_lbls = [r"$C_{E\parallel}(f)$", r"$C_{E\parallel}(f_{\text{fold}})$", 
            r"$C_{E\parallel}(\delta f)$"]
```

```python
from pyspedas import time_string
from matplotlib.ticker import MaxNLocator

fig, axes = plt.subplots(3, 3, figsize=(18, 14), dpi=300, constrained_layout=True)

cbar_label_fntsz = 18
axs_label_fntsz = 24
row_label_fntsz = 26
tick_label_fntsz = 14
cbar_ht = 0.05

# Figure title:
probe = 1
title_time = time_string(ntime[time_index])[:-3]
# fig.suptitle(f"MMS{probe} — {title_time}", fontsize=18, fontweight='bold')

pan_lbl = ['(a)', '(b)', '(c)']
for i in range(3):
    ax = axes[0, i]
    vmin, vmax = fpc.global_vmin_vmax_fpc(cfz_2d_list[i])
    ax1, ax2 = ax_pairs[i]
    _, ylabel = axlabel_pairs[i]
    height, width = len(ax2), len(ax1)
    im1, _ = fpc.imagecont_fpc(ax1, ax2, cfz_2d_list[i], xlabel=None, ylabel=ylabel, 
                               ax=ax, vmin=vmin, vmax=vmax, axis_fntsz=axs_label_fntsz)
    cbar = fig.colorbar(im1, ax=ax, fraction=0.047*height/width, pad=0.005)
    cbar.ax.set_title(r"$\text{W m}^{-3}$", fontsize=cbar_label_fntsz, pad=10)
    cbar.ax.tick_params(labelsize=tick_label_fntsz)
    axes[0,0].text(-0.45, 0.85, row_lbls[0], transform=axes[0,0].transAxes, 
                    va='center', ha='left', rotation='horizontal', fontsize=row_label_fntsz)
    ax.text(-0.2, 0.95, pan_lbl[i], transform=ax.transAxes, va='bottom', 
                    ha='left', fontsize=20, fontweight='bold')
    ax.tick_params(axis='both', labelsize=tick_label_fntsz)

    # Reduce axis ticks
    ax.xaxis.set_major_locator(MaxNLocator(nbins=3))
    ax.yaxis.set_major_locator(MaxNLocator(nbins=3))
    cbar.ax.yaxis.set_major_locator(MaxNLocator(nbins=5))

pan_lbl = ['(d)', '(e)', '(f)']
for j in range(3):
    ax = axes[1, j]
    vmin, vmax = fpc.global_vmin_vmax_fpc(cfoldz_2d_list[j])
    ax1, ax2 = ax_pairs[j]
    _, ylabel = axlabel_pairs[j]
    height, width = len(ax2), len(ax1)
    im1, _ = fpc.imagecont_fpc(ax1, ax2, cfoldz_2d_list[j], xlabel=None, ylabel=ylabel, 
                               ax=ax, vmin=vmin, vmax=vmax, axis_fntsz=axs_label_fntsz)
    cbar = fig.colorbar(im1, ax=ax, fraction=0.047*height/width, pad=0.005)
    cbar.ax.set_title(r"$\text{W m}^{-3}$", fontsize=cbar_label_fntsz, pad=10)
    cbar.ax.tick_params(labelsize=tick_label_fntsz)
    axes[1,0].text(-0.45, 0.85, row_lbls[1], transform=axes[1,0].transAxes, 
                    va='center', ha='left', rotation='horizontal', fontsize=row_label_fntsz)
    ax.text(-0.2, 0.95, pan_lbl[j], transform=ax.transAxes, va='bottom', 
                    ha='left', fontsize=20, fontweight='bold')
    ax.tick_params(axis='both', labelsize=tick_label_fntsz)

    # Reduce axis ticks
    ax.xaxis.set_major_locator(MaxNLocator(nbins=3))
    ax.yaxis.set_major_locator(MaxNLocator(nbins=3))
    cbar.ax.yaxis.set_major_locator(MaxNLocator(nbins=5))
    
    
pan_lbl = ['(g)', '(h)', '(i)']
for k in range(3):
    ax = axes[2, k]
    vmin, vmax = fpc.global_vmin_vmax_fpc(cz_2d_list[k])
    ax1, ax2 = ax_pairs[k]
    xlabel, ylabel = axlabel_pairs[k]
    height, width = len(ax2), len(ax1)
    im1, _ = fpc.imagecont_fpc(ax1, ax2, cz_2d_list[k], xlabel=xlabel, ylabel=ylabel, 
                               ax=ax, vmin=vmin, vmax=vmax, axis_fntsz=axs_label_fntsz)
    cbar = fig.colorbar(im1, ax=ax, fraction=0.047*height/width, pad=0.005)
    cbar.ax.set_title(r"$\text{W m}^{-3}$", fontsize=cbar_label_fntsz, pad=10)
    cbar.ax.tick_params(labelsize=tick_label_fntsz)
    axes[2,0].text(-0.45, 0.85, row_lbls[2], transform=axes[2,0].transAxes, 
                    va='center', ha='left', rotation='horizontal', fontsize=row_label_fntsz)
    ax.text(-0.2, 0.95, pan_lbl[k], transform=ax.transAxes, va='bottom', 
                    ha='left', fontsize=20, fontweight='bold')
    ax.tick_params(axis='both', labelsize=tick_label_fntsz)

    # Reduce axis ticks
    ax.xaxis.set_major_locator(MaxNLocator(nbins=3))
    ax.yaxis.set_major_locator(MaxNLocator(nbins=3))
    cbar.ax.yaxis.set_major_locator(MaxNLocator(nbins=5))

# Save routine
plt_pref = f'PRL_tripleFPC_{data_rate}_{species}_{bin_width_frac:.2f}'
pltfile = os.path.join(plot_dir, fpc.mms_name_make(plt_pref, 
                                                   trange[0], trange[1], ext='pdf'))
fig.savefig(pltfile, bbox_inches='tight', dpi=300)
print(f'File saved to {pltfile}!')

plt.show()
```
