```python
import numpy as np
import matplotlib.pyplot as plt
import os
import h5py
import fipcore as fpc
import pyspedas
from pyspedas.projects import mms
from pyspedas import get_data, tplot_rename, tplot_restore
```

```python
probe = '1'
trange = ['2015-12-09/05:03:55', '2015-12-09/05:03:59']
data_rate = 'brst'
level = 'l2'
species = 'e'
subtract_f0 = True
type_tag = 'df' if subtract_f0 else 'f'
bin_width_frac = 0.25
project_root = os.path.dirname(os.path.dirname(os.path.dirname(__file__))) if '__file__' in locals() else ".."
data_dir = os.path.join(project_root, "data")
plot_dir = os.path.join(project_root, "plots")
get_support_data = True
no_update = True
hfile_pref = f'mms{probe}_{data_rate}_{type_tag}_{species}_{bin_width_frac:.2f}'
hfile = os.path.join(data_dir, fpc.mms_name_make(hfile_pref, trange[0], trange[1]))
```

```python
hfile
```

```python
fpi_vars = mms.fpi(trange= trange, probe=probe, data_rate=data_rate, level=level,
        datatype=['des-dist', 'dis-dist', 'des-moms', 'dis-moms'], 
        time_clip=True, varnames=[f'mms{probe}_des_energy_brst', f'mms{probe}_des_phi_brst', 
        f'mms{probe}_des_bulkv_gse_brst', f'mms{probe}_dis_bulkv_gse_brst', 
        f'mms{probe}_des_temppara_brst', f'mms{probe}_des_tempperp_brst', 
        f'mms{probe}_des_numberdensity_brst', f'mms{probe}_des_dist_brst', 
        f'mms{probe}_des_disterr_brst', f'mms{probe}_des_errorflags_brst_dist'], 
        get_support_data=get_support_data, no_update=no_update)
```

```python
# Read in EDP data
edp_vars = mms.edp(trange=trange, probe=probe, data_rate=data_rate, level=level,
    datatype=['dce', 'scpot'],
    time_clip=True, varnames=[f'mms{probe}_edp_scpot_brst_l2', 
    f'mms{probe}_edp_dce_gse_brst_l2', f'mms{probe}_edp_dce_par_epar_brst_l2'], 
    get_support_data=get_support_data, no_update=no_update)
```

```python
# Read in FGM data
fgm_vars = mms.fgm(trange=trange, probe=probe, data_rate=data_rate, level=level,
    varnames=f'mms{probe}_fgm_b_gse_brst_l2', time_clip=True, 
    get_support_data=get_support_data, no_update=no_update)
```

```python
with h5py.File(hfile, "r") as f:
    ntime_e = f["meta"]["time"][...]
    bvec_lmn = f["lmn"]["bvec"][...]
    evec_lmn = f["lmn"]["evec_lor"][...]
    evec_fac = f["fac"]["evec_lor"][...]
    jvec_lmn = f["lmn"]["jvec"][...]
    jvec_fac = f["fac"]["jvec"][...]
    jdotE_fac = f["fac"]["jdotE"][...]
    jdotE_lmn = f["lmn"]["jdotE"][...]
    fac_mat = f["fac"]['rot_mat'][...]
    lmn_mat = f["lmn"]['rot_mat'][...]
    den_e = f['meta']['den_e'][...]
    den_i = f['meta']['den_i'][...]
    te_para = f['meta']['te_para'][...]
    te_perp = f['meta']['te_perp'][...]
    ti_para = f['meta']['ti_para'][...]
    ti_perp = f['meta']['ti_perp'][...]
    bulk_ve_lmn = f["lmn"]['bulk_ve'][...]
    bulk_ve_fac = f['fac']['bulk_ve'][...]
    bulk_vi_lmn = f["lmn"]['bulk_vi'][...]
    bulk_vi_fac = f["fac"]['bulk_vi'][...]
    bvec_lmn = f['lmn']['bvec'][...]
    evec_lmn = f['lmn']['evec_lor'][...]
    evec_fac = f['fac']['evec_lor'][...]
    jvec_lmn = f['lmn']['jvec'][...]
    jvec_fac = f['fac']['jvec'][...]
    jdotE_fac = f['fac']['jdotE'][...]
    jdotE_lmn = f['lmn']['jdotE'][...]
    bmag = f['meta']['bmag'][...]
```

```python
from pyspedas import time_datetime

t_ax = time_datetime(ntime_e)
axs_label_fntsz = 16   # adjust as needed
legend_fntsz = 14
legendFrame = True

n_subplots = 16

fig, ax = plt.subplots(n_subplots, 1, figsize=(12, 28),  sharex=True, constrained_layout=True)

# Figure title:
fig.suptitle(f"MMS {probe}: {trange[0]} – {trange[1]}", fontsize=18, fontweight='bold')

# First two subplots: energy spectra of electrons and ions
fig, ax0 = pyspedas.tplot('energy_e', fig = fig, axis = ax[0], display=False, return_plot_objects=True)
fig, ax1 = pyspedas.tplot('energy_i', fig = fig, axis = ax[1], display=False, return_plot_objects=True)
ax0.set_ylabel(r"$Electrons$" "\n" r"$(\mathrm{eV})$", fontsize=axs_label_fntsz)
ax1.set_ylabel(r"$Ions$" "\n" r"$(\mathrm{eV})$", fontsize=axs_label_fntsz)

# Third subplot: density of electrons and ions
ax[2].set_prop_cycle(color=['blue', 'red']) # custom color cycle
ax[2].plot(t_ax, den_e, linewidth=1.5, label=r'$n_e$')
ax[2].plot(t_ax, den_i, linewidth=1.5, label=r'$n_i$')
ax[2].axhline(0, color='k', linestyle='dotted', linewidth=1.0)
ax[2].set_ylabel(r"$n$" "\n" r"$(\mathrm{cm^{-3}})$", fontsize=axs_label_fntsz)
ax[2].legend(fontsize=legend_fntsz, frameon=legendFrame)

# Fourth subplot: electron temp profiles
ax[3].set_prop_cycle(color=['red', 'blue']) # custom color cycle
ax[3].plot(t_ax, te_para, linewidth=1.5, label=r'$T_{\parallel e}$')
ax[3].plot(t_ax, te_perp, linewidth=1.5, label=r'$T_{\perp e}$')
ax[3].axhline(0, color='k', linestyle='dotted', linewidth=1.0)
ax[3].set_ylabel(r"$T_e$" "\n" r"$(\mathrm{eV})$", fontsize=axs_label_fntsz)
ax[3].legend(fontsize=legend_fntsz, frameon=legendFrame)

# Fifth subplot: ion temp profiles
ax[4].set_prop_cycle(color=['red', 'blue']) # custom color cycle
ax[4].plot(t_ax, ti_para, linewidth=1.5, label=r'$T_{\parallel i}$')
ax[4].plot(t_ax, ti_perp, linewidth=1.5, label=r'$T_{\perp i}$')
ax[4].axhline(0, color='k', linestyle='dotted', linewidth=1.0)
ax[4].set_ylabel(r"$T_i$" "\n" r"$(\mathrm{eV})$", fontsize=axs_label_fntsz)
ax[4].legend(fontsize=legend_fntsz, frameon=legendFrame)

# Sixth subplot: electron bulk flow in lmn
ax[5].set_prop_cycle(color=['blue', 'green', 'red']) # custom color cycle
ax[5].plot(t_ax, bulk_ve_lmn, linewidth=1.5, label=["$U_{eL}$", "$U_{eM}$", "$U_{eN}$"])
ax[5].axhline(0, color='k', linestyle='dotted', linewidth=1.0)
ax[5].set_ylabel("$U_{eMVA}$" "\n" r"$(\mathrm{km/s})$", fontsize=axs_label_fntsz)
ax[5].legend(fontsize=legend_fntsz, frameon=legendFrame, loc='upper right')

# Seventh subplot: electron bulk flow in fac
ax[6].set_prop_cycle(color=['blue', 'green', 'red']) # custom color cycle
ax[6].plot(t_ax, bulk_ve_fac, linewidth=1.5, label=[r"$U_{e\perp_1}$", r"$U_{e\perp_2}$", 
                                                    r"$U_{e\parallel}$"])
ax[6].axhline(0, color='k', linestyle='dotted', linewidth=1.0)
ax[6].set_ylabel("$U_{eFAC}$" "\n" r"$(\mathrm{km/s})$", fontsize=axs_label_fntsz)
ax[6].legend(fontsize=legend_fntsz, frameon=legendFrame, loc='upper right')

# Eighth subplot: ion bulk flow in lmn
ax[7].set_prop_cycle(color=['blue', 'green', 'red']) # custom color cycle
ax[7].plot(t_ax, bulk_vi_lmn, linewidth=1.5, label=["$U_{iL}$", "$U_{iM}$", "$U_{iN}$"])
ax[7].axhline(0, color='k', linestyle='dotted', linewidth=1.0)
ax[7].set_ylabel("$U_{iMVA}$" "\n" r"$(\mathrm{km/s})$", fontsize=axs_label_fntsz)
ax[7].legend(fontsize=legend_fntsz, frameon=legendFrame, loc='upper right')

# Ninth subplot: ion bulk flow in fac
ax[8].set_prop_cycle(color=['blue', 'green', 'red']) # custom color cycle
ax[8].plot(t_ax, bulk_vi_fac, linewidth=1.5, label=[r"$U_{i\perp_1}$", r"$U_{i\perp_2}$", 
                                                    r"$U_{i\parallel}$"])
ax[8].axhline(0, color='k', linestyle='dotted', linewidth=1.0)
ax[8].set_ylabel(r"$U_{iFAC}$" "\n" r"$(\mathrm{km/s})$", fontsize=axs_label_fntsz)
ax[8].legend(fontsize=legend_fntsz, frameon=legendFrame, loc='upper right')

# Tenth subplot: B field in lmn
ax[9].set_prop_cycle(color=['blue', 'green', 'red', 'black']) # custom color cycle
ax[9].plot(t_ax, bvec_lmn, linewidth=1.5, label=['$B_L$', '$B_M$', '$B_N$'])
ax[9].plot(t_ax, bmag, linewidth=1.5, label='$|B|$')
ax[9].axhline(0, color='k', linestyle='dotted', linewidth=1.0)
ax[9].set_ylabel(r"$B$" "\n" r"$(\mathrm{nT})$", fontsize=axs_label_fntsz)
ax[9].legend(fontsize=legend_fntsz, frameon=legendFrame)

# Eleventh subplot: E field in LMN coordinates
ax[10].set_prop_cycle(color=['blue', 'green', 'red']) # custom color cycle
ax[10].plot(t_ax, evec_lmn, linewidth=1.5, label=["$E_L$", "$E_M$", "$E_N$"])
ax[10].axhline(0, color='k', linestyle='dotted', linewidth=1.0)
ax[10].set_ylabel(r"$E_{MVA}$" "\n" r"$(\mathrm{mV/m})$", fontsize=axs_label_fntsz)
ax[10].legend(fontsize=legend_fntsz, frameon=legendFrame, loc='upper right')

# Twelfth subplot: E field in FAC coordinates
ax[11].set_prop_cycle(color=['blue', 'green', 'red']) # custom color cycle
ax[11].plot(t_ax, evec_fac, linewidth=1.5, label=[r"$E_{\perp_1}$", r"$E_{\perp_2}$", 
                                                  r"$E_{\parallel}$"])
ax[11].axhline(0, color='k', linestyle='dotted', linewidth=1.0)
ax[11].set_ylabel("$E_{FAC}$" "\n" r"$(\mathrm{mV/m})$", fontsize=axs_label_fntsz)
ax[11].legend(fontsize=legend_fntsz, frameon=legendFrame, loc='upper right')

# Thirteenth subplot: J vector in lmn coordinates
ax[12].set_prop_cycle(color=['blue', 'green', 'red']) # custom color cycle
ax[12].plot(t_ax, jvec_lmn*1e-3, linewidth=1.5, label=["$J_L$", "$J_M$", "$J_N$"])
ax[12].axhline(0, color='k', linestyle='dotted', linewidth=1.0)
ax[12].set_ylabel(r"$J_{MVA}$" "\n" r"$(\mu\mathrm{A}/\mathrm{m}^2)$", fontsize=axs_label_fntsz)
ax[12].legend(fontsize=legend_fntsz, frameon=legendFrame, loc='upper right')


# Fourteenth subplot: J vector in fac coordinates
ax[13].set_prop_cycle(color=['blue', 'green', 'red']) # custom color cycle
ax[13].plot(t_ax, jvec_fac*1e-3, linewidth=1.5, label=[r"$J_{\perp_1}$", r"$J_{\perp_2}$", 
                                                       r"$J_{\parallel}$"])
ax[13].axhline(0, color='k', linestyle='dotted', linewidth=1.0)
ax[13].set_ylabel(r"$J_{FAC}$" "\n" r"$(\mu\mathrm{A}/\mathrm{m}^2)$", fontsize=axs_label_fntsz)
ax[13].legend(fontsize=legend_fntsz, frameon=legendFrame, loc='upper right')

# Fifteenth subplot: J.E in lmn coordinates
ax[14].set_prop_cycle(color=['black', 'blue', 'green', 'red']) # custom color cycle
ax[14].plot(t_ax, jdotE_lmn, linewidth=1.5, 
    label=[r"$J \cdot E_{tot}$", r"$J \cdot E_{L}$", r"$J \cdot E_{M}$", r"$J \cdot E_{N}$"])
ax[14].axhline(0, color='k', linestyle='dotted', linewidth=1.0)
ax[14].set_ylabel(r"$J \cdot E_{MVA}$" "\n" r"$(\mathrm{nW/m^3})$", fontsize=axs_label_fntsz)
ax[14].legend(fontsize=legend_fntsz, loc='upper left')

# Sixteenth subplot: J.E in fac coordinates
ax[15].set_prop_cycle(color=['black', 'blue', 'green', 'red']) # custom color cycle
ax[15].plot(t_ax, jdotE_fac, linewidth=1.5, 
    label=[r"$J·E_{tot}$", r"$J·E_{\perp_1}$", r"$J·E_{\perp_2}$", r"$J·E_{\parallel}$"])
ax[15].axhline(0, color='k', linestyle='dotted', linewidth=1.0)
ax[15].set_ylabel(r"$J \cdot E_{FAC}$" "\n" r"$(\mathrm{nW/m^3})$", fontsize=axs_label_fntsz)
ax[15].legend(fontsize=legend_fntsz, loc='upper left')


# Vertical reference lines
y_up = n_subplots + 1.36
ax[-1].axvline(x=dt_obj, ymin=0, ymax=y_up, color='k', linestyle='dashed', 
    linewidth=1, clip_on=False)
ax[-1].axvline(x=dt_obj0, ymin=0, ymax=y_up, color='k', linestyle='dotted', 
    linewidth=1, clip_on=False)
ax[-1].axvline(x=dt_obj1, ymin=0, ymax=y_up, color='k', linestyle='dotted', 
    linewidth=1, clip_on=False)

# Save routine
plt_pref = f'mms{probe}_{data_rate}_{species}_{bin_width_frac:.2f}'
pltfile = os.path.join(plot_dir, fpc.mms_name_make(plt_pref, 
                                                   trange[0], trange[1], ext='pdf'))
fig.savefig(pltfile, bbox_inches='tight', dpi=300)
print(f'File saved to {pltfile}!')

plt.show()
```

```python
import numpy as np
import h5py

def load_plt_mms_data(hfiles, time_indices):
    """
    Load FPC data for all probes and return integrated perp1, perp2, and parallel
    components for both C_E and dvdf, at a given time_index.
    """
    # If time_indices is just one int, convert to a list
    if isinstance(time_indices, int):
        time_indices = [time_indices] * len(hfiles)

    probes = []

    for hfile, time_index in zip(hfiles, time_indices):
        with h5py.File(hfile, "r") as f:
            c_vol = f["fac"]["c_vol"][...]          # (3,28,28,28,133)
            vpar   = f["fac"]["binc_n"][...]        # (28,)
            dvdf   = f["fac"]["bvdf_vol"][...]      # (28,28,28,133)
            tarr   = f["meta"]["time"][...]         # (133,)

        # Slice the 3 components of fpc and delta f at time_index
        c_perp1 = c_vol[0, :, :, :, time_index]
        c_perp2 = c_vol[1, :, :, :, time_index]
        c_par   = c_vol[2, :, :, :, time_index]

        dvdf_t = dvdf[:, :, :, time_index]

        # Integrate over two directions
        c_perp1_int = np.nansum(c_perp1, axis=(1, 2)) # over perp 2 and par
        c_perp2_int = np.nansum(c_perp2, axis=(0, 2)) # over perp 1 and par
        c_par_int   = np.nansum(c_par,   axis=(0, 1)) # over perp 1 and perp 2

        dvdf_perp1_int = np.nansum(dvdf_t, axis=(1, 2))
        dvdf_perp2_int = np.nansum(dvdf_t, axis=(0, 2))
        dvdf_par_int   = np.nansum(dvdf_t, axis=(0, 1))

        probes.append({
            "vpar": vpar,
            "c_e": [c_perp1_int, c_perp2_int, c_par_int],
            "df":  [dvdf_perp1_int, dvdf_perp2_int, dvdf_par_int],
            "time": tarr[time_index]
        })

    return probes
```

```python
import matplotlib.pyplot as plt
from pyspedas import time_string

def plt_vdf_fpc(trange, probes_data, time_index, prb=0, figsize=(18, 8), save_fig=False, 
             sav_ext='png'):
    """
    2×3 subplot layout:
      Row 0: delta f (df)
      Row 1: C_E
    """

    # vpar is already stored inside each probe dictionary
    vpar = probes_data[0]["vpar"]

    fig, axs = plt.subplots(2, 3, figsize=figsize, constrained_layout=True)

    # Style for MMS1–4
    colors  = ['red', 'green', 'blue', 'black']
    styles  = ['-', '--', ':', '-.']
    zorders = [5, 4, 3, 2]

    # Directly use probes_data — no extra dictionary needed
    for i, pdata in enumerate(probes_data):
        lbl = f"MMS{i+1}"
        color = colors[i]
        ls    = styles[i]
        zo    = zorders[i]

        for col in range(3):
            # Row 0: delta f
            axs[0, col].plot(
                vpar, pdata["df"][col],
                color=color, linestyle=ls, linewidth=2, zorder=zo, label=lbl
            )

            # Row 1: C_E
            axs[1, col].plot(
                vpar, pdata["c_e"][col],
                color=color, linestyle=ls, linewidth=2, zorder=zo, label=lbl
            )

    # Axis formatting
    for ax in axs.flat:
        ax.set_xlabel(r'$v_{\parallel}/v_{te}$', fontsize=20)
        ax.set_xlim([-3, 3])
        ax.tick_params(labelsize=16)
        ax.axhline(0, color='black', linewidth=1, linestyle='dashed')
        ax.axvline(0, color='black', linewidth=1, linestyle='dashed')
        ax.ticklabel_format(style='sci', axis='y', scilimits=(0, 0), useMathText=True)
        ax.yaxis.get_offset_text().set_fontsize(14)

    # Row labels
    for ax in axs[0, :3]:
        ax.set_ylabel(r'$\delta f\ (\mathrm{m^{-3}})$', fontsize=20)

    ce_labels = [r"C_{\mathrm{E}\perp 1}", r"C_{\mathrm{E}\perp 2}", r"C_{\mathrm{E}\parallel}"]
    for i, ax in enumerate(axs[1, :3]):
        ax.set_ylabel(rf'${ce_labels[i]}\ (\mathrm{{W\cdot m^{{-3}}}})$', fontsize=20)

    # Legend
    handles, labels = axs[0, 0].get_legend_handles_labels()
    fig.legend(handles, labels, loc='upper center', bbox_to_anchor=(0.5, 0.015),
               ncol=4, fontsize=16, frameon=True)

    # Annotation using probe's own stored time
    fig.text(0.98, 0.98, f"t: {time_index}", ha="right", va="top", fontsize=18)
    fig.suptitle(
        f"MMS: {time_string(probes_data[prb]['time'])[:-3]}",
        fontsize=18, fontweight='bold'
    )

    if save_fig:
        # Save routine
        plt_pref = f'1D_{time_index}_mms_{data_rate}_{species}_{bin_width_frac:.2f}'
        pltfile = os.path.join(plot_dir, fpc.mms_name_make(plt_pref, 
                                                trange[0], trange[1], ext=sav_ext))
        fig.savefig(pltfile, bbox_inches='tight', dpi=300)
        print(f'File saved to {pltfile}!')
        plt.close(fig)
    else:
        plt.show()
```

```python
probe = '1'
trange = ['2015-12-09/05:03:55', '2015-12-09/05:03:59']
# trange = ['2015-12-09/05:03:56.5', '2015-12-09/05:03:57.2']
data_rate = 'brst'
level = 'l2'
species = 'e'
subtract_f0 = True
type_tag = 'df' if subtract_f0 else 'f'
bin_width_frac = 0.25
project_root = os.path.dirname(os.path.dirname(os.path.dirname(__file__))) if '__file__' in locals() else ".."
data_dir = os.path.join(project_root, "data")
plot_dir = os.path.join(project_root, "plots")
get_support_data = True
no_update = True
hfile_pref = f'mms{probe}_{data_rate}_{type_tag}_{species}_{bin_width_frac:.2f}'
hfile = os.path.join(data_dir, fpc.mms_name_make(hfile_pref, trange[0], trange[1]))
```

```python
with h5py.File(hfile, "r") as f:
    ntime_e = f["meta"]["time"][...]
    bvec_lmn = f["lmn"]["bvec"][...]
    bvec_gse = f["gse"]["bvec"][...]
```

```python
from pyspedas import time_datetime

t_ax = time_datetime(ntime_e)
axs_label_fntsz = 16   # adjust as needed
legend_fntsz = 14
legendFrame = True

n_subplots = 2

fig, ax = plt.subplots(n_subplots, 1, figsize=(12, 4),  sharex=True, constrained_layout=True)

# Figure title:
fig.suptitle(f"MMS {probe}: {trange[0]} – {trange[1]}", fontsize=18, fontweight='bold')

# First subplot bvec_gse
ax[0].set_prop_cycle(color=['blue', 'green', 'red']) # custom color cycle
ax[0].plot(t_ax, bvec_gse , linewidth=1.5, label=['$B_x$', '$B_y$', '$B_z$'])
ax[0].set_prop_cycle(color=['black', 'black', 'black']) # custom color cycle
ax[0].axhline(0, color='k', linestyle='dotted', linewidth=1.0)
ax[0].set_ylabel(r"$B$" "\n" r"$(\mathrm{nT})$", fontsize=axs_label_fntsz)
ax[0].legend(fontsize=legend_fntsz, frameon=legendFrame)

# Second subplot bvec_lmn
ax[1].set_prop_cycle(color=['blue', 'green', 'red']) # custom color cycle
ax[1].plot(t_ax, bvec_lmn , linewidth=1.5, label=['$B_L$', '$B_M$', '$B_N$'])
ax[1].axhline(0, color='k', linestyle='dotted', linewidth=1.0)
ax[1].set_ylabel(r"$B$" "\n" r"$(\mathrm{nT})$", fontsize=axs_label_fntsz)
ax[1].legend(fontsize=legend_fntsz, frameon=legendFrame)

# Vertical reference lines
y_up = n_subplots + 0.087
idx = 61
dt_obj = time_datetime(ntime_e[idx])
ax[-1].axvline(x=dt_obj, ymin=0, ymax=y_up, color='k', linestyle='dashed', 
    linewidth=1, clip_on=False)

# Save routine
plt_pref = f'mms{probe}_{data_rate}_{species}_{bin_width_frac:.2f}'
pltfile = os.path.join(plot_dir, fpc.mms_name_make(plt_pref, 
                                                   trange[0], trange[1], ext='pdf'))
fig.savefig(pltfile, bbox_inches='tight', dpi=300)
print(f'File saved to {pltfile}!')

plt.show()
```
