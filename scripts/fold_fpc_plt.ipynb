```python
import numpy as np
from fipcore import mms_name_make, slice3d_to_2d, imagecont_vdf_fpcfold
import os
```

```python
trange = ['2015-12-09/05:03:55', '2015-12-09/05:03:59']
```

```python
species = 'e' 
bin_width_frac = 0.25
h5_path = os.path.join("..", "data")
hfile_pref = f'mms_fpc_{species}_{bin_width_frac:.2f}'
hfile = os.path.join(h5_path, mms_name_make(hfile_pref, trange[0], trange[1]))
hfile
```

```python
trange = ['2015-12-09/05:03:55', '2015-12-09/05:03:59']
t_list = [58, 59, 60, 61, 62, 63, 64]
for t_ind in t_list:
    imagecont_vdf_fpcfold(trange, species='e', bin_width_frac=0.25, time_index = t_ind,
            coord_type='fac', suptitle_sfx='', dpi_val=100, axis_fntsz=16, 
            cbar_ht=0.050, save_fig=False, outdir_sfx='', show_tind=True)
    print(f'Files saved for time_index = {t_ind}')
```
